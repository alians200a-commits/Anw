```python
from kokoro import KPipeline
from IPython.display import display, Audio, Markdown
import numpy as np
import soundfile as sf
from pathlib import Path

pipeline = KPipeline(lang_code='a')
voices = ['af_heart', 'af_bella', 'af_sarah']
words = ['Propofol', 'Suxamethonium', 'Rocuronium']
out = Path('/content/kokoro_samples')
out.mkdir(exist_ok=True)

for voice in voices:
    display(Markdown(f'## {voice}'))
    for word in words:
        chunks = []
        for _, _, audio in pipeline(word, voice=voice, speed=0.95):
            chunks.append(audio)
        audio = np.concatenate(chunks) if len(chunks) > 1 else chunks[0]
        path = out / f'{voice}-{word.lower()}.wav'
        sf.write(path, audio, 24000)
        print(word)
        display(Audio(audio, rate=24000))
```

```python
import shutil
shutil.make_archive('/content/kokoro_voice_samples', 'zip', out)
print('جاهز: /content/kokoro_voice_samples.zip')
```
